# 01 · Crear el cache

Decodifica todos los DICOM de train y guarda series `uint8` de 256×256. Lo corre Rodrigo una sola vez; puede tardar varias horas, usa CPU y mucho disco, no necesita GPU. Los demás solo usan el resultado.

```python
# CONFIG
RUN_NAME = "cache_v1_256"
RAW_ROOT = "/workspace/data/raw"
CACHE_VERSION = "v1_256"
IMAGE_SIZE = 256
N_WORKERS = 12
```

```python
from pathlib import Path
import sys

repo = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "src" / "knee").exists())
sys.path.insert(0, str(repo))
from src.knee.cache import build_cache

raw = Path(RAW_ROOT)
cache = Path("/workspace/data/cache") / CACHE_VERSION
print(f"Entrada: {raw / 'train_series'}")
print(f"Salida: {cache}")
```

```python
index = build_cache(
    raw / "train_series.csv",
    raw / "train_series",
    cache,
    image_size=IMAGE_SIZE,
    workers=N_WORKERS,
)
display(index.head())
display(index.groupby("plane").agg(series=("series", "count"), cortes=("n_slices", "sum")))
```
